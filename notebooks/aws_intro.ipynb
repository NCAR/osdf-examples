## AWS EC2 via CloudBank

- Amazon Web Services (AWS) EC2 provides on-demand virtual machines in the cloud. NSF-funded researchers can obtain AWS compute through [CloudBank](https://www.cloudbank.org/) by exchanging an [ACCESS](https://allocations.access-ci.org/) allocation for cloud credits; CloudBank then provisions an AWS account. See the [CloudBank documentation](https://www.cloudbank.org/training) for the current steps.
- The datasets used in these notebooks live in AWS Open Data buckets in the **us-west-2** region (e.g. `cmip6-pds`, `ncar-cesm2-lens`), so launch your EC2 instance in **us-west-2**. Same-region access is fast, avoids egress charges, and makes the OSDF-vs-S3 comparison fair.
- Launch an Ubuntu instance (`m5.xlarge` or `m5.4xlarge` suit these workflows) with an SSH key pair and inbound SSH (port 22) allowed from your IP.

:::{note}: Connecting and setting up JupyterLab
Connect and forward JupyterLab to your laptop over SSH:

    ssh -i your-key.pem -L 8888:localhost:8888 ubuntu@<public-dns>

Create a Python environment from this repository's `requirements.txt` (a virtual environment or a conda env), then start JupyterLab:

    jupyter lab --no-browser --port 8888

Open the printed `http://localhost:8888/lab?token=...` URL in your browser — the SSH tunnel makes it resolve to the instance.
:::
